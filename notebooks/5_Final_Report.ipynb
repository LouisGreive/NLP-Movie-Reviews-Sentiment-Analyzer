# NLP Movie Review Sentiment Analyzer
## Final Report

This notebook summarizes the final results of the end-to-end NLP project. three different models were built/deployed and evaluated on the full 50,000-review IMBD dataset to determine the best architecture for binary sentiment classification.



## 1. Final Performance Results

The table below shows the definitive weighted average F1-scores and Accuracy**¹** for all three models after being trained and tested on the full dataset. 
The F1-score is the primary metric as it provides a balanced measure of precision and recall. 

| Model                       | Accuracy | F1-Score (Weighted Avg) |
| --------------------------- | -------- | ----------------------- |
| Baseline (TF-IDF + N-grams) | 0.89     | 0.89                    |
| Simple NN (Word Vectors)    | 0.77     | 0.77                    |
| State-of-the-Art (Transformer) | 0.89     | 0.89                    |

**¹ The equality between the Accuracy and the weighted average F1-score is a mathematical property observed in this specific case due to the test dataset being perfectly balanced (equal number of positive and negative samples).**

## 2. Final Analysis and Conclusion 

The final result of this project are fascinating. Trough the experimentation with three different model architectures,it was possible to test the designed solutions and deploy finally the best performing one.

The main takeaway from the project is that the **baseline model**, created using TF-IDF with N-grams on a cleaned dataset, performs at the same level as the **transformer model** with both achieving a final F-1 Score of **0.89** and an Accuracy of 0.89. The Simple neural network using Word vectors underformed both with a F1-Score of **0.77**. 

The conclusion is that for a well defined task like binary sentiment analysis, a robust and well engineeered classic model can be just as powerful as a more complex deep learning architecture, especially when trained on a large data-set (When trained on the 1.000-review sample the baseline model underperfomed the Transformer model, achieving a F-1 score of only **0.70**) The N-grams architecture was highly effective at capturing the key phrases needed for classification. 

While the transformer has a more nuanced understanding of the language and the context, for this specific task, the baseline model proved to be more efficient and equally effective solution. The approach of starting with a with a simple baseline model for comparative purposes was highly valuable,  as the baseline itself ended up being a model with excellent performance despite its lower level of complexity. 



## Visual Performance Comparison 

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Data for the plot with final scores
model_names = ['Baseline (TF-IDF)', 'Simple NN (Word Vectors)', 'Transformer']
f1_scores = [0.89, 0.77, 0.89] 

plt.figure(figsize=(10, 6))
sns.barplot(x=f1_scores, y=model_names)
plt.title('Final Model F1-Score Comparison (Full Dataset)', fontsize=16)
plt.xlabel('F1-Score (Weighted Average)', fontsize=12)
plt.xlim(0, 1.0)

for index, value in enumerate(f1_scores):
    plt.text(value, index, f' {value:.2f}', va='center')
plt.show()
```
