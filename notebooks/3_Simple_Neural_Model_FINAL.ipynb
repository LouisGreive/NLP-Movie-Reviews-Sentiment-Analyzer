# Advanced Neural Model 

## Objective
In this notebook, I will build an simple neural model to challenge the strong TF-IDF baseline Pipeline.
Using the pre-trained **Word Vectors**, which captures the semantic meaning of text, my hypothesis is that the model will better understand linguist nuance and achive a higher F1-Score.


## 1.Setup and Load
First I will import the necessary libraries and load the cleaned,preprocessed dataset from the previous notebook  (`1_Data_Exploration_and_Cleaning.ipynb`).

```python
import pandas as pd 
import numpy as np
import spacy
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import classification_report

# Loading of the FULL cleaned data
file_path = '../data/reviews_cleaned_FULL.csv'
df_clean = pd.read_csv(file_path)

# Drop any empty rows just in case
df_clean.dropna(inplace=True)

print('Cleaned data loaded successfully.')
```

## Creation of the Word Vectors
The core of the advanced approach: Instead of counting words (like TF-IDF) I will represent each review as a single, dense, 300-dimension **document vector** using the pre-trained 'en_core_web_md' model from'spaCy'. 
These Vectors (Embeddings) represent the "meaning" of the text, allowing the model to understand context and semantic relations between words

```python
# Load the medium spaCy model
print("Loading spaCy model...")
nlp = spacy.load('en_core_web_md')
print('Model loaded.')

# Get the vector for each review on the cleaned review column 
# The .vector attribute gives the 300-dimension semantic representation
# Stack all into a single Numpy array for the model
x_vectors = np.array([nlp(review).vector for review in df_clean['cleaned_review']])
y = df_clean['sentiment']

print('All reviews has been converted to vectors.')
```

## Split of the Dataset
Just as with the baseline, I will split the new vector-based dataset into an 80% training set and 20% testing set. 
This ensures a fair an unbiased evaluation on data that the model has never seen before

```python
# Split the vector data
x_train_vectors, x_test_vectors, y_train, y_test = train_test_split(x_vectors, y, test_size =0.2, random_state=42, stratify=y)

print(f"Training set size: {len(x_train_vectors)}")
print(f"Testing set size: {len(x_test_vectors)}")
```

## Model training
I will use a Multi-layer Perceptron('MLPClassifier') from scikit-learn. This is a simple but powerful feed-forward neural network architecture. It will learn to map the 300-dimension input vectors from the reviews to a final "positive" or "negative" sentiment prediction.

```python
# Initialize the neural network (multi-layer Perceptron Classifier)
# Set max_iter to a higher number to give it more time to learn
nn_model = MLPClassifier(random_state=42, max_iter=500)

print('Training the Advanced Neural Model...')
nn_model.fit(x_train_vectors, y_train)
print('Training complete!')
```

## Evaluation of the simple neural model
I will use the trained neural network to make a prediction on the unseen data. I will generate a full classification report to compare its performance directly against the baseline's F1-score of 0.71.

```python
# Make predictions on the unseen test data
y_pred_nn = nn_model.predict(x_test_vectors)

# the final report
print('\n Simple Neural model Perfomance')
print(classification_report(y_test, y_pred_nn))
```

## Findings and Conclusion (Full dataset)


### Results summary
After training on the full 40.000 review training set, the simple neural model using word vectors and a simple Neural Network achieved a **weighted average F1-score of 0.77**

While it is a good perfomance, it is significantly lower than the **0.89 F1-score** achieved by the TF-IDF N-gram baseline model.

## Analysis of the result
This is a valuable and realistic finding that highlights a key concept in NLP. 

While the **Word Vectors** provide a richer semantic representation of the text, than TF-IDF, the simple MLPClassifier architecture may not have been complex enough to fully exploit this richness.

The baseline's success demonstrates that its use of **N-grams** was a highly effective strategy for this specific task, allowing to capture key sentiment phrases directly. 

## Justification for the final model.
The underperformance of this model, even on the full dataset, provides a powerful justification for the final stage of this project. It is clear that to unlock the full potencial of semantic word vectors, a more sophisticated, context-aware neural architecture is required.
